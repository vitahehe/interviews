# Interview theme-grid explorer

This notebook loads the generated theme-grid JSON files and displays each interview's themes in a readable format for qualitative review.

```python
from pathlib import Path
import json
from IPython.display import Markdown, display

repo_root = Path.cwd()
if not (repo_root / 'make_themes').exists():
    repo_root = repo_root.parent

json_dir = repo_root / 'make_themes' / 'outputs' / 'interview_theme_grids' / 'json'
json_files = sorted(json_dir.glob('*.json'))
json_files
```

```python
for path in json_files:
    print(path.name)
```

```python
selected = json_files[0] if json_files else None
if selected is None:
    display(Markdown('No theme-grid JSON files were found.'))
else:
    payload = json.loads(selected.read_text(encoding='utf-8'))
    display(Markdown(f'## {payload.get("interview_id", selected.stem)}'))
    for theme in payload.get('theme_grid', []):
        display(Markdown(f'### {theme.get("theme_name", "")}'))
        display(Markdown(theme.get('theme_description', '')))
        for code in theme.get('codes', []):
            examples = '\n'.join([f'- {e.get("participant_id", "")}: {e.get("quote", "")}' for e in code.get('examples', [])])
            display(Markdown(f'- **{code.get("code_name", "")}**: {code.get("description", "")}\n  - Motivation: {code.get("motivation", "")}\n{examples}'))
```
