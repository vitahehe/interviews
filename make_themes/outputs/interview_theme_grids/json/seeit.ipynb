```python
import json
import pandas as pd
from pathlib import Path
from IPython.display import display, HTML

# -------------------------------------------------------------------
# 1. Path to your theme-grid JSON output
# -------------------------------------------------------------------
# Change this path if your file is elsewhere.
json_path = Path("/home/vita/projects/interviews/make_themes/outputs/interview_theme_grids/json/Monday_20260223_Kvant1_lab_transcript.json")

# If you are testing with the uploaded file in this ChatGPT sandbox, use:
# json_path = Path("/mnt/data/Pasted code.json")

# -------------------------------------------------------------------
# 2. Load JSON
# -------------------------------------------------------------------
with open(json_path, "r", encoding="utf-8") as f:
    data = json.load(f)

print("Interview ID:", data.get("interview_id", "UNKNOWN"))
print("Status:", data.get("status", "API/parsed JSON"))
print()

# -------------------------------------------------------------------
# 3. Flatten theme grid into table rows
# -------------------------------------------------------------------
rows = []

for theme in data.get("theme_grid", []):
    theme_label = f"{theme.get('theme_id', '')} {theme.get('theme_name', '')}".strip()
    theme_description = theme.get("theme_description", "")

    codes = theme.get("codes", [])

    for i, code in enumerate(codes):
        examples = code.get("examples", [])

        example_texts = []
        for ex in examples:
            participant = ex.get("speaker_label_original") or ex.get("participant_id", "")
            chunk_id = ex.get("chunk_id", "")
            quote = ex.get("quote", "")
            theory_codes = ex.get("existing_theory_codes", [])
            why = ex.get("why_example_fits", "")

            theory_codes_str = ", ".join(theory_codes)

            example_texts.append(
                f"<b>{participant}</b> "
                f"<span style='color:gray;'>({chunk_id})</span>: "
                f"“{quote}”"
                f"<br><span style='color:#666;'><i>Existing codes:</i> {theory_codes_str}</span>"
                f"<br><span style='color:#666;'><i>Why it fits:</i> {why}</span>"
            )

        examples_html = "<br><br>".join(example_texts)

        # Only show theme name on first row of each theme group
        rows.append({
            "Themes": theme_label if i == 0 else "",
            "Theme description": theme_description if i == 0 else "",
            "Codes": f"{code.get('code_id', '')} {code.get('code_name', '')}".strip(),
            "Description": code.get("description", ""),
            "Motivation": code.get("motivation", ""),
            "Examples": examples_html,
            "Confidence": code.get("confidence", ""),
            "Human review needed": code.get("human_review_needed", "")
        })

theme_grid_df = pd.DataFrame(rows)

# -------------------------------------------------------------------
# 4. Display as readable HTML table
# -------------------------------------------------------------------
html = theme_grid_df.to_html(
    escape=False,
    index=False
)

styled_html = f"""
<style>
table {{
    border-collapse: collapse;
    width: 100%;
    font-family: Arial, sans-serif;
    font-size: 14px;
}}
th {{
    background-color: #f2f2f2;
    text-align: left;
    padding: 8px;
    border: 1px solid #ddd;
    vertical-align: top;
}}
td {{
    padding: 8px;
    border: 1px solid #ddd;
    vertical-align: top;
}}
td:nth-child(1) {{
    font-weight: bold;
    min-width: 180px;
}}
td:nth-child(3) {{
    font-weight: bold;
    min-width: 180px;
}}
td:nth-child(6) {{
    min-width: 350px;
}}
</style>
<h2>Theme grid: {data.get("interview_id", "UNKNOWN")}</h2>
{html}
"""

display(HTML(styled_html))

# -------------------------------------------------------------------
# 5. Optional: display overlooked candidates and warnings
# -------------------------------------------------------------------
overlooked = data.get("overlooked_candidates", [])
if overlooked:
    display(HTML("<h2>Overlooked candidate themes/codes</h2>"))
    overlooked_df = pd.DataFrame(overlooked)
    display(overlooked_df)

case_notes = data.get("case_level_notes", {})
if case_notes:
    display(HTML("<h2>Case-level notes</h2>"))

    for key, value in case_notes.items():
        display(HTML(f"<h3>{key}</h3>"))
        if isinstance(value, list):
            for item in value:
                display(HTML(f"<li>{item}</li>"))
        else:
            display(HTML(str(value)))
```
